# NumPy Foundations

## 1. Why do arrays exist?

```python
import numpy as np

ages = np.array([19, 21, 20, 22, 18])

print(ages)
print(type(ages))
```

```python
print(ages + 1)
print(ages * 2)
print(ages - 5)
print(ages / 2)
```

Normal Python List 

```python
ages_list = [19, 21, 20, 22, 18]

print(ages_list + [1])
print(ages_list * 2)
```

```python
ages + 1
```

```python
ages + np.array([1, 1, 1, 1, 1])
```

```python
ages + np.array([1, 2, 3, 4, 5])
```

What I understand now: NumPy arrays allow numerical operations to be applied element-by-element, unlike Python lists where + means concatenation.

```python
print(ages.ndim)
print(ages.shape)
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 88, 72],
    [92, 81, 95]
])
print(marks)
print(marks.ndim)
print(marks.shape)
```

```python
print(marks[0])
print(marks[1])
print(marks[2])
```

```python
print(marks[0, 0])
print(marks[1, 2])
print(marks[2, 1])
```

```python
marks[:, 0]
```

```python
marks[0, :]
```

```python

marks[:2, :]
```

```python
marks[:, :2]
```

```python
marks + 5
```

```python
marks * 2
```

```python
marks[:, 0] + 10
```

```python
print(marks)
```

```python
marks_list = [
    [80, 75, 90],
    [65, 88, 72],
    [92, 81, 95]
]

# Normal Python approach
result = []

for row in marks_list:
    new_row = []
    for mark in row:
        new_row.append(mark + 5)
    result.append(new_row)

print(result)
```
